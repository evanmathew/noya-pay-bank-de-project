# FX Rates - Bronze

- Parse `rate_date` as a date and `rate` as `decimal(18,6)`
- Set rates that are not positive to null

**Reads:** `bronze.bronze_fx_rates`  
**Writes:** `silver.silver_fx_rates` (overwrite)

## Imports

```python
from pyspark.sql import functions as F
```

## Read bronze table

```python
fx_rate_df = spark.read.table("bronze.bronze_fx_rates")
display(fx_rate_df)
```

## Clean and type the columns

```python
fx_rate = (
    fx_rate_df.withColumn("rate_date", F.to_date("rate_date"))
    .withColumn("rate", F.col("rate").cast("decimal(18,6)"))
    .withColumn("rate", F.when(F.col("rate") > 0, F.col("rate")))
)

display(fx_rate)
```

## Write to silver

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS silver")
fx_rate.write.mode("overwrite").saveAsTable("silver.silver_fx_rates")
```
